```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def tanh(x):
    return math.tanh(x)

X = [1, 2, 3]

Wf, Whf, bf = 0.5, 0.1, 0
Wi, Whi, bi = 0.6, 0.2, 0
Wc, Whc, bc = 0.7, 0.3, 0
Wo, Who, bo = 0.8, 0.4, 0

Wy, by = 4, 0

h = 0.0
C = 0.0

print("LSTM Numerical Example")
print("Input sequence:", X)
print("Initial h0 =", h, "C0 =", C)
print()

for t, x in enumerate(X, start=1):

    f = sigmoid(Wf * x + Whf * h + bf)
    i = sigmoid(Wi * x + Whi * h + bi)
    C_tilde = tanh(Wc * x + Whc * h + bc)

    C = f * C + i * C_tilde

    o = sigmoid(Wo * x + Who * h + bo)

    h = o * tanh(C)

    print("Time Step", t)
    print("f =", round(f,3))
    print("i =", round(i,3))
    print("C_tilde =", round(C_tilde,3))
    print("C =", round(C,3))
    print("o =", round(o,3))
    print("h =", round(h,3))
    print()

x4 = 4

f4 = sigmoid(Wf * x4 + Whf * h + bf)
i4 = sigmoid(Wi * x4 + Whi * h + bi)
C_tilde4 = tanh(Wc * x4 + Whc * h + bc)
C4 = f4 * C + i4 * C_tilde4
o4 = sigmoid(Wo * x4 + Who * h + bo)
h4 = o4 * tanh(C4)

print("Time Step 4 Prediction")
print("f4 =", round(f4,3))
print("i4 =", round(i4,3))
print("C_tilde4 =", round(C_tilde4,3))
print("C4 =", round(C4,3))
print("o4 =", round(o4,3))
print("h4 =", round(h4,3))
print()

y_hat = Wy * h4 + by

print("Final Prediction")
print("y_hat =", round(y_hat,3))
```
